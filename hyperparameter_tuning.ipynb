```python
import os
from ultralytics import YOLO

# ================= CONFIGURATION =================
# Path to your data.yaml file
DATA_YAML_PATH = r"path/to/your/dataset/data.yaml"  # <--- UPDATE THIS

# Directory where results will be saved
PROJECT_DIR = "YOLO_Tuning_Runs"
TUNING_NAME = "My_Tuning_Experiment"

# Model weights (Make sure you have this file locally or it will try to download)
MODEL_WEIGHTS = "yolo26m.pt" 

# ================= EXECUTION =================
if __name__ == "__main__":
    print(f"🚀 Starting Hyperparameter Tuning for {MODEL_WEIGHTS}...")
    
    # Initialize the model
    model = YOLO(MODEL_WEIGHTS)

    # Run the tuning
    # This will run multiple short training sessions to find the best settings
    model.tune(
        data=DATA_YAML_PATH,
        project=PROJECT_DIR,
        name=TUNING_NAME,
        
        # Tuning Settings
        epochs=30,       # Train for 30 epochs per experiment
        iterations=30,   # Run 30 different experiments (Higher = better results but takes longer)
        
        # Fixed Hyperparameters (We don't want to tune these)
        optimizer='MuSGD', # Fixing optimizer as requested
        
        # Operational Settings
        plots=False,     # Turn off plotting for speed
        save=False,      # Don't save checkpoint for every tuning run (saves space)
        val=True,        # Validate during tuning to get mAP scores
        gpu=0            # Use GPU 0 (Change if you have multiple GPUs)
    )

    print("\n🎉 TUNING COMPLETE!")
    print(f"Check '{PROJECT_DIR}/{TUNING_NAME}' for the 'best_hyperparameters.yaml' file.")
```
